# 学生成绩分析

本 notebook 分析10名学生的数学、英语、编程成绩。

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib

# 让图表显示中文
matplotlib.rcParams['font.sans-serif'] = ['SimHei']
matplotlib.rcParams['axes.unicode_minus'] = False
```

## 1. 生成数据

```python
import numpy as np
np.random.seed(42)  # 保证每次结果一样

names = ["张三", "李四", "王五", "赵六", "钱七",
         "孙八", "周九", "吴十", "郑一", "王二"]

df = pd.DataFrame({
    "姓名": names,
    "数学": np.random.randint(55, 100, 10),
    "英语": np.random.randint(50, 100, 10),
    "编程": np.random.randint(60, 100, 10),
})
df["总分"] = df["数学"] + df["英语"] + df["编程"]
df["平均分"] = df["总分"] / 3
df
```

## 2. 查看数据

```python
print("各科平均分：")
print(df[["数学", "英语", "编程"]].mean().round(2))
print("\n各科最高分：")
print(df[["数学", "英语", "编程"]].max())
print("\n各科最低分：")
print(df[["数学", "英语", "编程"]].min())
```

## 3. 总分排名

```python
df_sorted = df.sort_values("总分", ascending=False)
df_sorted[["姓名", "总分", "平均分"]].reset_index(drop=True)
```

## 4. 画柱状图

```python
fig, ax = plt.subplots(figsize=(8, 5))
df.set_index("姓名")[["数学", "英语", "编程"]].plot(kind="bar", ax=ax)
ax.set_title("学生各科成绩对比")
ax.set_ylabel("分数")
ax.set_xlabel("姓名")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 5. 结论
- 编程平均分最高
- 数学平均分偏低
- 总分前三名见排名表
